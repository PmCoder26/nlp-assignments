```python
import nltk

nltk.download('punkt')
nltk.download('wordnet')
```

```python
text = "Hello! This is an example sentence. Let's test tokenization, stemming, and lemmatization. #NLP is fun  "
```

Whitespace-based Tokenization

```python
from nltk.tokenize import WhitespaceTokenizer

wt = WhitespaceTokenizer()
wt.tokenize(text)
```

Punctuation-based Tokenization

```python
from nltk.tokenize import wordpunct_tokenize

wordpunct_tokenize(text)
```

Treebank Tokenization

```python
from nltk.tokenize import TreebankWordTokenizer

twt = TreebankWordTokenizer()
twt.tokenize(text)
```

Tweet Tokenization

```python
from nltk.tokenize import TweetTokenizer

tt = TweetTokenizer()
tt.tokenize(text)
```

MWE Tokenization

```python
from nltk.tokenize import MWETokenizer

mwe = MWETokenizer([('New', 'York'), ('NLP', 'is')])
text2 = "New York is my home and NLP is"
mwe.tokenize(text2.split())
```

Porter Stemmer

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()

words = wt.tokenize(text)

print("Porter Stemming:")
for word in words:
    print(word, "->", ps.stem(word))
```

Snowball Stemmer

```python
from nltk.stem import SnowballStemmer

sbs = SnowballStemmer('english')

print('Snoball Stemming: ')
for word in words:
    print(word, '->', sbs.stem(word))
```

WordNet Lemmatization

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

print("Lemmatization:")
for w in words:
    print(w, "->", lemmatizer.lemmatize(w))
```
